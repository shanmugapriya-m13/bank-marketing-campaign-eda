### CUSTOMERS ANALYSIS

```python
#Age Distribution
plt.figure(figsize = (10,5))
sns.histplot(Bank_TD['age'],
             bins = 30,
             kde=True,
             color = 'steelblue')
plt.title("Age Distribution of Bank Customers")
plt.xlabel("Age")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
#Job Type Breakdown
plt.figure(figsize=(10,5))
sns.countplot(data = Bank_TD, x = 'job',
              hue = 'job',
              order = Bank_TD['job'].value_counts().index,
              palette = 'Blues_d')
plt.title("Customer Count by Job Type",
          fontsize = 16,
          fontweight = "bold")
plt.xlabel("Job",
           fontsize = 16,
           fontweight = "bold")
plt.ylabel("Count",
           fontsize = 16,
           fontweight = "bold")
plt.xticks(rotation = 45,
           fontsize = 12)
plt.tight_layout()
plt.show()
```

```python
#Marital Status Breakdown
plt.figure(figsize=(7,5))
sns.countplot(data = Bank_TD,
              x = 'marital',
              hue = 'marital',
              palette = 'Set2')
plt.title("Marital Status of Customers")
plt.xlabel("Marital Status")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

### SUBSCRIPTION ANALYSIS

```python
#Overall Subscription Rate
plt.figure(figsize=(6,4))
sns.countplot(data=Bank_TD,
              x='y',
              hue = 'y',
              palette = 'Set1')
plt.title("Overall Subscription Count (Yes = Subscribed)")
plt.xlabel("Subscribed or Not")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
print(Bank_TD['y'].value_counts(normalize=True) * 100)
```

```python
#Subcription Rate by Job
plt.figure(figsize=(12,5))
sns.countplot(data = Bank_TD,
              x = 'job',
              hue = 'y',
              order = Bank_TD['job'].value_counts().index,
              palette = 'Set1')
plt.title("Subscription Rate by Job Type",
           fontsize = 15,
           fontweight ='bold')
plt.xlabel("Job",
            fontsize = 15,
            fontweight ='bold')
plt.ylabel("Count",
           fontsize = 15,
           fontweight ='bold')
plt.xticks(rotation = 45,
           fontsize = 12)
plt.legend(title = "Subscribed",
           title_fontsize = 15,
           fontsize =13)
plt.tight_layout()
plt.show()
```

```python
#Subscription Rate by Education
plt.figure(figsize=(12,5))
sns.countplot(data = Bank_TD,
              x = 'education',
              hue = 'y',
              palette = 'Set2')
plt.title("Subscription Rate by Education Level")
plt.xlabel("Education")
plt.ylabel("Count")
plt.xticks(rotation = 30)
plt.legend(title = "Subscribed",
           title_fontsize = 15)
plt.tight_layout()
plt.show()
```

```python
#Creating Age bins
Bank_TD['Age_Group'] = pd.cut(Bank_TD['age'],
                              bins = [17,30,45,60,100],
                              labels = ['18-30', '31-45','46-60','60+'],
                              include_lowest = True)
```

```python
#Subscription Rate by Age Group
plt.figure(figsize = (8,5))
sns.countplot(data = Bank_TD,
              x = 'Age_Group',
              hue = 'y',
              palette = 'coolwarm')
plt.title("Subscription Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Count")
plt.legend(title="Subscribed")
plt.tight_layout()
plt.show()
```

```python
#Call Duration vs Subscription
plt.figure(figsize=(8,5))
sns.boxplot(data = Bank_TD,
            x = 'y',
            hue = 'y',
            y = 'duration',
            palette = 'Set3',
            legend = False)
plt.title("Call Duration VS Subscription Outcome")
plt.xlabel("Subscribed to Term Deposit?")
plt.ylabel("Call Duration (Seconds)")
plt.tight_layout()
plt.show()

#Average durtion for each group
print(Bank_TD.groupby('y')['duration'].mean())
```

```python
#Campaign Contact Patterns
#How Many Times was Each Customer Contacted?
plt.figure(figsize=(10, 5))
sns.histplot(Bank_TD['campaign'],
             bins=20,
             color='coral')
plt.title("Number of Contacts During Campaign",
          fontsize = 13)
plt.xlabel("Number of Contacts",
           fontsize = 13)
plt.ylabel("Count",
           fontsize = 13)
plt.tight_layout()
plt.show()

# Print Average Contacts
print("Average contacts per customer:", Bank_TD['campaign'].mean().round(2))

#Did Previous Campaign Outcome Affect this One?
plt.figure(figsize=(8, 5))
sns.countplot(data=Bank_TD,
              x='poutcome',
              hue='y',
              palette='Set1')
plt.title("Previous Campaign Outcome vs Current Subscription")
plt.xlabel("Previous Outcome")
plt.ylabel("Count")
plt.legend(title="Subscribed")
plt.tight_layout()
plt.show()
```

Summary of Key Findings

1. Admin workers appear most in the dataset, but students and retired 
customers had a better subscription rate relative to their group size.

2. The 31-45 age group had the highest number of subscriptions by count, 
but younger (18-30) and older (60+) customers subscribed at a higher rate 
proportionally.

3. Call duration was notably different between the two groups — customers 
who subscribed spent an average of around 550 seconds on the call, compared 
to around 220 seconds for those who did not. This suggests that genuine 
interest drives longer conversations.

4. Most customers were contacted between 1 to 3 times during the campaign. 
Contacting customers more than 5 times showed no meaningful improvement in 
subscription, which means repeated follow-ups are not an effective strategy.

5. The majority of customers had no prior campaign history. However, among 
those who did, a successful previous outcome was the strongest indicator of 
subscribing again — making past customers a valuable target group.
